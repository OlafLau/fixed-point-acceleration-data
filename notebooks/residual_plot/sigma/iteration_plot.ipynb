```python
import mpltex
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.offsetbox import OffsetImage, AnnotationBbox
from matplotlib.cbook import get_sample_data
import pandas as pd
import numpy as np
import random, itertools, math, re, os
import csv
from collections import defaultdict
os.environ["PATH"] = "/Library/TeX/texbin:" + os.environ["PATH"]
```

```python
plt.rcParams['pgf.texsystem'] = 'pdflatex'
plt.rcParams['text.latex.preamble'] = r'\usepackage{amsmath}'
```

```python
df = pd.read_csv('../plot_algorithms.csv')
```

```python
@mpltex.acs_decorator
def plot_sd_iterations():
    fig, ax = plt.subplots(figsize=(3.3,2.2))
    line_style = mpltex.linestyle_generator(lines=['-'], markers=[], hollow_styles=[])
    line_style_list = [next(line_style) for _ in range(10)]
    
    sd_variants = ['AA_SD', 'OA_SD', 'NA_SD']
    labels = ['AA-SD', 'OA-SD', 'NA-SD']
    
    for idx, updater in enumerate(sd_variants):
        for _, r in df.iterrows():
            if r['updater'] == updater:
                res = eval(r['residuals'])
                iterations = np.arange(1, len(res) + 1)
                style = line_style_list[idx % len(line_style_list)]
                ax.plot(iterations, res, label=labels[idx], **style)
                break
    
    ax.set_xlabel('iterations', fontsize=10)
    ax.set_yscale('log')
    plt.yticks([1e-11, 1e-9, 1e-7, 1e-5, 1e-3, 1e-1]) 
    ax.legend(loc='best')
    ax.grid(False)
    plt.tight_layout(pad=0.1)
    plt.savefig('SI_sd_iterations_plot.svg')
    plt.show()
plot_sd_iterations()
```
